```python
from tqdm import tqdm # Progress bar
from PIL import Image

# Base output directory
output_dir = r"custom_data"

classes = ["Glioma", "Pituitary", "Meningioma",]

# Create folder structure
for split in ["train", "val", "test"]:
    for cls in classes:
        os.makedirs(os.path.join(output_dir, split, cls), exist_ok=True)

# Helper function to save images
def save_images(dataset, split_name, size=(224, 224)):
    for idx, example in enumerate(tqdm(dataset, desc=f"Saving {split_name} images")):
        label_idx = example['label']         # integer label
        label = classes[label_idx]           # map to class name
        image = example['image']             # PIL Image object

        # Resize to 224x224
        image = image.resize(size, resample=Image.LANCZOS)
        
        # Destination path
        filename = f"{split_name}_{idx}.png" # unique filename
        dst_path = os.path.join(output_dir, split_name, label, filename)

        # Save image
        image.save(dst_path)

# Save images
save_images(train_dataset, "train")
save_images(valid_dataset, "val")

print("✅ Images saved successfully!")
```
